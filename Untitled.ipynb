```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sc
import math

#%% 
#numerical scheme #Solving Diffusion Equation with different BC
def Theta_MethodDiffusionEquation(a,b,c,d,T,N,u0,beta,gamma,theta):
    '''
    Solves the 1D Diffusion Equation for u(x,t) using the following boundary conditions:
    u(x,0) = u0, 
    ux + alpha*u = beta, for x=0 
    ub = u(b,t)
    with rectangular mesh(x,t) N by T (T is not the final time, T+1 is the number of nodes!)
    theta takes inputs between 0,1 corresponding to different finite element schemes
    '''
    #Setting up space-time mesh. time is columns (tmax = last col) and spcae is rows(last row is xmax)
    if theta < 0 or theta > 1:
        return print("Theta must be in [0,1]")
    dx = (b-a)/N 
    dt = (d-c)/T
    time = np.array([c+m*dt for m in range(T+1)])
    space = np.array([a+dx*m for m in range(N+1)])
    mu = dt / (dx)**2 
    #Setting up my entire U
    U = np.zeros((N+1,T+1))
 
    #Initial and Boundary conditions
    u_xinitial = np.array(u0(space))
    U[:,0] = u_xinitial

    #boundaries AT x=0 are enforced within the scheme
   
    #Building A matrix 
    A =np.zeros((N+1,N+1)) +np.diag(np.array([-2 for m in range(N+1)]),k=0) +np.diag(np.array([1 for m in range(N)]),k=1)+np.diag(np.array([1 for m in range(N)]),k=-1)
    A1 = np.identity(N+1)- mu* theta* A # we are solving a system of the type A1*U^m+1 = A2*U^m + g
    A2 = np.identity(N+1) + mu* (1-theta) * A
    #Finding the first 
    A1[0,1] = 1/dx
    A1[-1,:] = 0
    A1[-1,-1] =1/dx
    A1[-1,-2] = -1/dx
    #A2 depends only on lower timestep
    A2[0,:] = 0
    A2[-1,:] = 0
    #for neumann, need to make last tow 
    for i in range(T):
        A1[0,0] = beta(time[i+1]) - 1/dx
        #bc2 = ub(time[i])
        #gm = np.zeros(N-1);gm[0] = bc1;gm[-1] = bc2
        gm1   = np.zeros(N+1);gm1[0] = gamma(time[i+1])
        b = A2 @ U[:,i] + gm1
        U[:,i+1] = np.linalg.solve(A1,b)
    #An idea for Current# First order
    I = (U[1,:]-U[0,:])/ dx 
    #Second Order Current Option:since u0=0
    I2 = (-3*U[0,:]+4*U[1,:]-U[2,:])/(2*dx)
    return U,I,I2
#%%
# %%
#Initial Conditions
a= 0 #x0
b=10 #xmax
c= 0 #t_initial
d= 40 #t_final
N = 100 #x-mesh
T = 10000 #t-mesh
u0 = lambda x: 1 #u(x,0)
Estart = -10
E0 = 0
trev = 20
k0 = 25
alpha = 0.5
dE = 0.1
omega = np.pi
E_lin = lambda t: Estart + t
E_lincyc = lambda t: (Estart + t)*(t<=trev) + (Estart + 2*trev - t)*(t>trev)
E_ac = lambda t: E_lin(t) + dE*np.sin(omega*t) 
E_ac_cyc = lambda t: E_lincyc(t) + dE*np.sin(omega*t)
E = lambda t: E_lin(t) #use this to change type of potential
gamma = lambda t: -k0*np.exp(-alpha*(E(t) - E0))
beta = lambda t: -k0*np.exp(-alpha*(E(t)-E0)) * (np.exp(E(t)-E0) + 1)
theta= 1
sol,I_neumann,I2_neumann= Theta_MethodDiffusionEquation(a,b,c,d,T,N,u0,beta,gamma,theta)
#%%
#plotting concentration at some time t
#Plot concentration at final time:
x = np.linspace(a,b,N+1)
plt.plot(x, sol[:, 2000], label=r"Implicit Euler")
plt.xlabel("x")
plt.ylabel(f"u(x, t={d:.3f})")
plt.legend()
plt.show()# %%
#exact solution
# %%
t =np.linspace(c,d,T+1)
xx,tt = np.meshgrid(x,t)
#contour 
fig,ax = plt.subplots()
CS = ax.contour(xx,tt,sol.T)
ax.clabel(CS)
# %%
#plot current wrt time:
#Avoid division by zero in the exact solution by introducing epsilon
# %%
```

```python
import numpy as np
from numpy import pi
from scipy.integrate import cumulative_trapezoid
from scipy.sparse import diags, csr_matrix
from scipy.sparse.linalg import spsolve
import matplotlib.pyplot as plt

# ----------------------------
# User-defined E(t), E_dot(t)
# ----------------------------
# Replace these with your actual E(t) and E'(t).
def E(t):
    return E_lin(t)
def E_dot(t):
    return 1

# ----------------------------
# Problem parameters
# ----------------------------
L = 10.0
E0 = 0.7

# Time grid
Tfinal = 1e-03
Nt = 2000
t = np.linspace(0.0, Tfinal, Nt)
dt = t[1] - t[0]

# h(t) and kappa
h = 1.0 / (1.0 + np.exp(E(t) - E0))
kappa = 1.0 - h[0]

# sech^2 helper
def sech2(z):
    return 1.0 / np.cosh(z)**2

# g(t) = -h'(t) = (E_dot/4)*sech^2((E-E0)/2)
g = (E_dot(t) / 4.0) * sech2((E(t) - E0) / 2.0)

# ----------------------------------------
# Series evaluation of a_x(0,t) from modes
# ----------------------------------------
def ax0_series(t, g, kappa, L, Nterms=200):
    """
    Returns a_x(0,t) using Nterms of the eigen-expansion.
    Uses the form:
      Phi_n(t)=exp(-k_n^2 t)*(kappa*b_n + b_n*∫_0^t g(s)*exp(k_n^2 s) ds)
      a_x(0,t)=Σ k_n*Phi_n(t)
    """
    Nt = len(t)
    ax0 = np.zeros(Nt, dtype=float)

    for n in range(Nterms):
        k_n = (n + 0.5) * np.pi / L
        b_n = 4.0 / ((2*n + 1) * np.pi)

        exp_kn2_t = np.exp((k_n**2) * t)
        # J_n(t) = ∫_0^t g(s) exp(k_n^2 s) ds
        integrand = g * exp_kn2_t
        Jn = cumulative_trapezoid(integrand, t, initial=0.0)

        Phi = np.exp(-(k_n**2) * t) * (kappa * b_n + b_n * Jn)

        ax0 += k_n * Phi

    return ax0

# Compute series current (spatial derivative)
Nterms = 300
ax0_ana = ax0_series(t, g, kappa, L, Nterms=Nterms)


```
